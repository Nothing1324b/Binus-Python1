```python
def write_biodata():
  name = input("Enter Your Name: ")
  age = input("Enter Your Age: ")
  address = input("Enter Your Address: ")
  email = input("Enter Your Email: ")

  with open("Biodata.txt", "w") as file:
    file.write(f"Name: {name}\n")
    file.write(f"Age: {age}\n")
    file.write(f"Address: {address}\n")
    file.write(f"Email: {email}\n")
  print("\nBiodata successfully saved to Biodata.txt!")


def read_biodata():
  print("\n--- Reading Biodata.txt ---")
  with open("Biodata.txt", "r") as file:
    content = file.read()
    print(content)



write_biodata()
read_biodata()
```

```python
def file_create(filename):
    text = input("Enter text: ")
    with open(filename, "w") as file:
        file.write(text + "\n")
    print("File has been made!")


def read_file(filename):
    try:
        with open(filename, "r") as file:
            print(file.read())
    except FileNotFoundError:
        print("Error: File not found.")


def append_file(filename):
    text = input("Enter text to add: ")
    with open(filename, "a") as file:
        file.write(text + "\n")
    print("Text has been added!")


actions = {
    "1": file_create,
    "2": read_file,
    "3": append_file
}

while True:
    print("1. Create file")
    print("2. Read file")
    print("3. Append file")
    print("4. Close")

    choice = input("Choose (1/2/3/4): ")

    if choice == "4":
        print("see ya!")
        break

    action = actions.get(choice)
    if action:
        filename = input("File name: ")
        action(filename)
    else:
        print("Invalid choice")
```

```python
import pandas as pd

# Read the CSV file into a DataFrame
df = pd.read_csv('countries_data.csv')

# Display the first few rows to verify the data
print("Country Data:")
print(df)
print("-" * 40)

# Calculate and display the Mean and Standard Deviation for numeric columns
print("Mean (Average):")
print(df.mean(numeric_only=True))
print("-" * 40)

print("Standard Deviation:")
print(df.std(numeric_only=True))
```
